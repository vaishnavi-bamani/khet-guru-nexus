## Notebook 02 : Exploratory Data Analysis (EDA)

### 🎯 Objective

The objective of this notebook is to explore the Crop Recommendation dataset through visualization.

This notebook helps us understand:

- Feature distributions
- Crop distribution
- Relationships between variables
- Correlation among features
- Outliers
- Insights useful for machine learning

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../datasets/crop-recommendation/Crop_recommendation.csv")
```

```python
crop_counts = df["label"].value_counts()

plt.figure(figsize=(12,6))
crop_counts.plot(kind="bar")
plt.title("Distribution of Crop Labels")
plt.xlabel("Crop")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

### Observation

The dataset contains multiple crop categories.

The distribution appears balanced, which is beneficial for training a classification model because it reduces class imbalance.

```python
df.hist(figsize=(15,10))
plt.tight_layout()
plt.show()
```

```python
corr = df.drop(columns=["label"]).corr()

plt.figure(figsize=(8,6))
plt.imshow(corr)

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.colorbar()

plt.title("Correlation Matrix")

plt.show()
```

```python
features = df.columns[:-1]

for feature in features:
    plt.figure(figsize=(5,3))
    plt.boxplot(df[feature])
    plt.title(feature)
    plt.show()
```

# 📊 Insights from EDA

From the exploratory analysis we observed:

- Dataset is balanced.
- Features have different numerical ranges.
- Some variables contain potential outliers.
- Most relationships are weak to moderate.
- Dataset appears suitable for machine learning.
