# 🌾 Data Understanding

In this section, we load the Crop Recommendation dataset and perform an initial inspection to understand its structure and quality.

The analysis includes:

- Importing the required libraries
- Loading the dataset
- Displaying the first five records
- Viewing dataset information
- Generating statistical summary
- Listing all feature names
- Checking for missing values

```python
# Import Libraries
import pandas as pd
import numpy as np

# Load Dataset
df = pd.read_csv("../datasets/crop-recommendation/Crop_recommendation.csv")

# Initial Data Inspection
print("=" * 60)
print("FIRST FIVE ROWS")
print("=" * 60)
print(df.head())

print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)
df.info()

print("\n" + "=" * 60)
print("STATISTICAL SUMMARY")
print("=" * 60)
print(df.describe())

print("\n" + "=" * 60)
print("COLUMN NAMES")
print("=" * 60)
print(df.columns)

print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)
print(df.isnull().sum())
```

# 📌 Dataset Quality Check

Before training any machine learning model, it is important to verify that the dataset is clean and suitable for analysis.

In this section, we check:

- Dataset dimensions

```python
print("=" * 60)
print("DATASET SHAPE")
print("=" * 60)
print(df.shape)

print("\n" + "=" * 60)
print("DUPLICATE ROWS")
print("=" * 60)
print(df.duplicated().sum())

print("\n" + "=" * 60)
print("TARGET VARIABLE")
print("=" * 60)
print(df["label"].value_counts())
```
